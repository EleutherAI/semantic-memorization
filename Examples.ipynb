```python
dataset = dataset['memories_deduped_12b']
```

```python
dataset = dataset.to_pandas()
```

```python
def get_model_taxonomy(row):
    if row.sequence_duplicates >= 6:
        return "recitation"
    elif row.is_incrementing or row.is_repeating:
        return "reconstruction"
    else:
        return "recollection"
```

```python
dataset['model_taxonomy'] = dataset.apply(get_model_taxonomy, axis=1)
```
